NLTK قدم اول: بارگذاری پیکره از طریق کتابخانه پردازش متن



```python
import nltk
from nltk.corpus import reuters

nltk.download('reuters')

doc_ids = reuters.fileids()
print("Number of documents:", len(doc_ids))


text = reuters.raw(doc_ids[0])
print("\nText of the document:",doc_ids[0],"\n", text)  # print first 500 chars

documents=[(doc_id,reuters.raw(doc_id)) for doc_id in doc_ids]
```

قدم دوم: توکن سازی اسناد

```python
from nltk import word_tokenize
nltk.download('punkt_tab')

documents_tokenized=[(doc_id,word_tokenize(text)) for doc_id,text in documents]

print(documents_tokenized[0:20])
```
